```python
from sklearn.linear_model import LinearRegression
X = [[100],[150],[200],[250],[300]] #training data
y = [50,75,100,125,150] #labels

model = LinearRegression()
model.fit(X,y) #training the model

print(model.predict([[200]]))
```
